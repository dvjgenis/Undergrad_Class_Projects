# 2. Displaying the Images
Below are the three images of the more iconic landscapes and features of each location, with relative paths used.

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

images = ['Dawn_Charles_V_Palace_Alhambra_Granada_Andalusia_Spain.jpg', 'Hero_Best-Skyline-Views.jpg', 'palawan-philippines-coron-cr-alamy.jpg']

for imgfile in images:
    img = Image.open(imgfile)

    img_data = np.array(img)

    fig, ax = plt.subplots(1, 1)
    ax.imshow(img_data)
    plt.show()
```

# 3. Color Distribution Analysis (Histograms)
Below, after perfecting my redundant code using for-loops, I have displayed the color distibution histograms of each location accordingly. All are unique and varied, illustrating the diversity and distinctiveness of these locations.

```python
for img_file in images:
    img = Image.open(img_file)
    img_data = np.array(img) 
    
    r, g, b = img_data[:,:,0], img_data[:,:,1], img_data[:,:,2]

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].hist(r.ravel(), bins=256, color="r")
    axes[0].set_title("Red Histogram")

    axes[1].hist(g.ravel(), bins=256, color="g")
    axes[1].set_title("Green Histogram")

    axes[2].hist(b.ravel(), bins=256, color="b")
    axes[2].set_title("Blue Histogram")

    plt.suptitle(f"Color Distribution: {img_file}")
    plt.show()
```

# 4. Image Color Distribution Overview Analysis

### Image 2: Chicago Skyline

### Image 3: Palawan, Philippines
