**Example Code:** (A simple sine wave plot)

```python
# Week 1: Basic Plot with Matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Generate data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Plot the sine wave
plt.figure(figsize=(8, 4))
plt.plot(x, y, label='sin(x)')
plt.title("Introduction to Data Visualization: Sine Wave")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

## Week 4: Interactivity with Widgets

**Topics Covered:**
- Adding interactivity using widgets and traitlets.
- Interactive data binding.

**Key Concepts:**
- ipywidgets for dynamic plots.
- Enhancing user engagement with interactive controls.

**Example Code:** (Interactive sine wave plot using a slider)

```python
# Week 4: Interactive Plot with ipywidgets
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact

def update_plot(freq=1):
    x = np.linspace(0, 10, 200)
    y = np.sin(freq * x)
    plt.figure(figsize=(8, 4))
    plt.plot(x, y)
    plt.title(f"Sine Wave with Frequency = {freq}")
    plt.xlabel("X-axis")
    plt.ylabel("sin(x)")
    plt.show()

interact(update_plot, freq=(0.5, 5.0, 0.5))
```

## Week 5: Grammar of Graphics & bqplot

**Topics Covered:**
- Introduction to bqplot and the grammar of graphics.
- Building interactive charts.

**Key Concepts:**
- Declarative charting with bqplot.
- Understanding the components of a visualization (scales, axes, marks).

**Example Code:** (A simple bqplot bar chart)

```python
# Week 5: bqplot Bar Chart Example
import bqplot as bq
import numpy as np

# Define sample data for a bar chart
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 12, 37]

# Define scales
x_sc = bq.OrdinalScale()
y_sc = bq.LinearScale()

# Create a bar chart
bars = bq.Bars(x=categories, y=values, scales={'x': x_sc, 'y': y_sc})
ax_x = bq.Axis(scale=x_sc, label='Category', orientation='horizontal')
ax_y = bq.Axis(scale=y_sc, label='Value', orientation='vertical')

chart = bq.Figure(marks=[bars], axes=[ax_x, ax_y], title='bqplot Bar Chart Example')
chart
```

**Example Code:** (An interactive Altair bar chart)

```python
# Week 8: Altair Interactive Chart Example
import altair as alt
import pandas as pd

# Sample data for Altair
data = pd.DataFrame({
    'category': ['A', 'B', 'C', 'D'],
    'value': [28, 55, 43, 91]
})

chart = alt.Chart(data).mark_bar().encode(
    x='category',
    y='value',
    tooltip=['category', 'value']
).properties(title="Altair Bar Chart Example")
chart
```

## Week 12: Advanced Streamlit & HuggingFace

**Topics Covered:**

**Key Concepts:**
- Building more complex and interactive web apps.
- Data validation and presentation.

**Example Code:** (An advanced Streamlit app with Altair integration)

```python
# Week 12: Advanced Streamlit Interaction Example
import streamlit as st
import pandas as pd
import altair as alt

st.title("Advanced Streamlit App")

# Create a sample DataFrame
df = pd.DataFrame({
    'x': range(10),
    'y': [i ** 2 for i in range(10)]
})

st.write("Quadratic Data")
st.dataframe(df)

# Generate an Altair chart
chart = alt.Chart(df).mark_line().encode(
    x='x',
    y='y'
).properties(title="Quadratic Function")

st.altair_chart(chart, use_container_width=True)
```

## Week 13: 3D Graphics & Scientific Visualization

**Topics Covered:**
- Introduction to 3D graphics in visualization.
- Basics of scientific visualization techniques.

**Key Concepts:**
- 3D data representation.
- Utilizing libraries like Plotly for interactive 3D plots.

**Example Code:** (3D scatter plot using Plotly)

```python
# Week 13: 3D Scatter Plot with Plotly
import plotly.graph_objects as go

# Generate random 3D data
np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100)
z = np.random.randn(100)

fig = go.Figure(data=[go.Scatter3d(
    x=x, y=y, z=z,
    mode='markers',
    marker=dict(
        size=5,
        color=z,  # Color by z value
        colorscale='Viridis',
        opacity=0.8
    )
)])
fig.update_layout(title="3D Scatter Plot Example")
fig.show()
```

**Example Code:** (Generating a word cloud)

```python
# Week 16: Word Cloud Example
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = (
    "Data visualization is an essential tool for understanding complex data patterns "
    "and communicating insights effectively."
)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Word Cloud Example")
plt.show()
```
