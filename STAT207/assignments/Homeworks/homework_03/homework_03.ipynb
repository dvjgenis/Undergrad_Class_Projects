## 3. Exploring Athlete Weights [2 points]

**a)** First, generate a graphical summary (histogram or boxplot) of the weight variable.

```python
df['Weight'].hist()
plt.title('Histogram of Olympic Weight Variable')
plt.xlabel('Weight')
plt.ylabel('Frequency')
plt.show()
```

```python
df['Weight'].plot(kind='box')
plt.title('Boxplot of Olympic Weight Variable')
plt.ylabel('Weight')
plt.show()
```
